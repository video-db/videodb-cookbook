```python
import os
from getpass import getpass

from IPython.display import Audio, Image
from videodb import connect

os.environ["VIDEO_DB_API_KEY"] = getpass("Please enter your VideoDB API Key: ")

conn = connect(base_url="https://api.dev.videodb.io")
coll = conn.get_collection()

print("Connected to VideoDB successfully.")
```

## 3. Generate an Image

Use `generate_image()` to create an image from a text prompt.

```python
image_prompt = "A cinematic neon jellyfish floating through a futuristic green-lit aquarium"

generated_image = coll.generate_image(
    prompt=image_prompt,
    aspect_ratio="9:16",
)

image_url = generated_image.generate_url()
print("Image URL:", image_url)
Image(url=image_url)
```

## 4. Generate Music

Use `generate_music()` to create background music from a prompt.

```python
music_prompt = "Upbeat electronic background music for a product launch video"

generated_music = coll.generate_music(
    prompt=music_prompt,
    duration=10,
)

music_url = generated_music.generate_url()
print("Music URL:", music_url)
Audio(url=music_url, filename="music.mp3")
```

## 5. Generate a Sound Effect

Use `generate_sound_effect()` for short, scene-specific audio assets.

```python
sfx_prompt = "Footsteps on wet gravel in a quiet alley, realistic and slightly echoey"

generated_sfx = coll.generate_sound_effect(
    prompt=sfx_prompt,
    duration=5,
)

sfx_url = generated_sfx.generate_url()
print("Sound effect URL:", sfx_url)
Audio(url=sfx_url, filename="sound-effect.mp3")
```

## 6. Generate a Voiceover

Use `generate_voice()` to turn text into spoken audio.

```python
voiceover_text = (
    "Welcome to VideoDB. In this demo, we are generating images, music, "
    "sound effects, voiceovers, videos, and translated media from simple prompts."
)

generated_voice = coll.generate_voice(
    text=voiceover_text,
    voice_name="Aria",
)

voice_url = generated_voice.generate_url()
print("Voiceover URL:", voice_url)
Audio(url=voice_url, filename="voiceover.mp3")
```

## 7. Generate a Short Video

Use `generate_video()` to create a short video clip from a prompt.

```python
video_prompt = (
    "Cinematic close-up of a majestic lion slowly turning its head, "
    "golden mane catching soft afternoon sunlight on the savanna."
)

generated_video = coll.generate_video(
    prompt=video_prompt,
    duration=7,
)

generated_video.play()
```

## 8. Search YouTube

Use `youtube_search()` to discover videos that can be uploaded or used as source material in your workflow.

```python
search_query = "learn python programming"

youtube_results = conn.youtube_search(
    query=search_query,
    result_threshold=3,
    duration="medium",
)

for i, result in enumerate(youtube_results, 1):
    print(f"{i}. {result.get('title', 'Untitled')} - {result.get('link', '')}")
```

## 9. Translate a Video Transcript

This section uses the new understanding pipeline.

We will:

1. Upload a source video.
2. Run `understand()` with the `spoken_words` analyzer.
3. Convert the transcript artifact into plain text.
4. Translate the transcript with `generate_text()`.

Because we only need transcript text for generation, no `video.index()` call is required.

```python
source_video = coll.upload(url="https://www.youtube.com/watch?v=a9__D53WsUs")
source_video.play()
```

```python
transcript_understanding = source_video.understand(
    analyzers=[
        {
            "type": "spoken_words",
            "name": "transcript",
            "config": {
                "language": "en",
            },
        }
    ],
)

transcript_understanding.wait_until_complete(timeout=3600, poll_interval=15)
transcript_output = transcript_understanding.get_analyzer("transcript").get_output()

print("Transcript status:", transcript_output.get("status"))
print("Transcript scenes:", len(transcript_output.get("scenes", [])))
```

```python
def build_transcript_text(transcript_output):
    text_parts = []

    for scene in transcript_output.get("scenes", []):
        data = scene.get("data", {})
        scene_text = data.get("text")

        if not scene_text:
            words = data.get("words", [])
            scene_text = " ".join(word.get("text", "") for word in words)

        scene_text = scene_text.strip()
        if scene_text:
            text_parts.append(scene_text)

    return "\n".join(text_parts)


transcript_text = build_transcript_text(transcript_output)
print("Transcript characters:", len(transcript_text))
print(transcript_text[:1000])
```

```python
def extract_generated_text(response):
    if isinstance(response, dict):
        return response.get("output") or response.get("text") or str(response)
    return str(response)


target_language = "French"
translation_prompt = f"""
Translate the following transcript into {target_language}.
Preserve paragraph breaks where helpful. Return only the translated transcript.

Transcript:
{transcript_text[:14000]}
"""

translated_transcript = extract_generated_text(
    coll.generate_text(prompt=translation_prompt, model_name="pro")
).strip()

print(translated_transcript)
```

## 10. Dub an Existing Video

Use `dub_video()` when you want VideoDB to create a dubbed version of a video in another language.

```python
dub_source_video = coll.upload(url="https://www.youtube.com/watch?v=FgrO9ADPZSA")
target_language_code = "hi"

dubbed_video = coll.dub_video(
    video_id=dub_source_video.id,
    language_code=target_language_code,
)

print("Dubbed video ID:", dubbed_video.id)
dubbed_video.play()
```

You now have a compact toolkit for generating and transforming media with VideoDB:

- Images, music, sound effects, voiceovers, and videos from prompts.
